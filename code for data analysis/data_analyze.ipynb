```python
fs_dir = fetch_fsaverage(verbose=True)
subjects_dir = op.dirname(fs_dir)
subject = 'fsaverage'
trans = 'fsaverage'  # MNE has a built-in fsaverage transformation
snr = 3.0
lambda2 = 1.0 / snr ** 2
src = mne.setup_source_space(subject, spacing='ico4', add_dist='patch',subjects_dir=subjects_dir)
conductivity = (0.3, 0.006, 0.3)  # for three layers
model = mne.make_bem_model(subject=subject, ico=3,conductivity=conductivity,subjects_dir=subjects_dir)
bem = mne.make_bem_solution(model)
```

```python
def stc(eeg,eeg_w):
    fwd = mne.make_forward_solution(eeg.info, trans=trans, src=src, bem=bem,meg=False, eeg=True, mindist=5.0, n_jobs=1,verbose=True)
    inv = mne.minimum_norm.make_inverse_operator(eeg.info, fwd, eeg_w)
    stc_SCE_reg = mne.minimum_norm.apply_inverse_epochs(eeg, inv, lambda2, 'dSPM')
    del fwd,inv,
    return stc_SCE_reg
def stc_evoked(eeg,eeg_w):
    fwd = mne.make_forward_solution(eeg.info, trans=trans, src=src, bem=bem,meg=False, eeg=True, mindist=5.0, n_jobs=1,verbose=True)
    inv = mne.minimum_norm.make_inverse_operator(eeg.info, fwd, eeg_w)
    stc_SCE_reg = mne.minimum_norm.apply_inverse(eeg, inv, lambda2, 'dSPM')
    del fwd,inv,
    return stc_SCE_reg
```

```python
first_result = epochs['0003','0004','0005']
first_result_w = epochs_w['0003','0004','0005']
first_result_evoked = first_result.average()
first_result_evoked.set_eeg_reference('average',projection=True)
first_result_cov = mne.compute_covariance(first_result_w)
first_result_stc = stc_evoked(first_result_evoked,first_result_cov)
del first_result,first_result_w
```
