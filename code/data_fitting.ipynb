```python
class RL_agent():
    def __init__(self,gamma=0.25,LR=0.5):
# 0:stay-cue; 1:con1(high); 2:con2(low) safe-risk
        # self.Q = np.zeros((2,3)) + np.ones((2,3))*6
        self.Q = np.array([[6,6,6],[6,6,6]])
        self.lr = LR
        self.gamma = gamma

    def P_action_softmax(self,state,action):
        # Q = np.matmul(self.Q,state)
        Q = self.Q[:,0]
        exp_Q = np.sum(np.exp(Q*self.gamma))
        return (np.exp(self.gamma*Q)/exp_Q)[action]
    
    def P_action_con1(self):
        Q = self.Q[:,1]
        exp_Q = np.sum(np.exp(Q*self.gamma))
        return np.exp(Q*self.gamma)/exp_Q
    
    def P_action_con2(self):
        Q = self.Q[:,2]
        exp_Q = np.sum(np.exp(Q*self.gamma))
        return np.exp(Q*self.gamma)/exp_Q
        
    def P_action_con0(self):
        Q1 = (self.Q[0,1]+self.Q[0,2])/2
        Q2 = (self.Q[1,1]+self.Q[1,2])/2
        Q = np.array([Q1,Q2])
        exp_Q = np.sum(np.exp(Q*self.gamma))
        return np.exp(Q*self.gamma)/exp_Q
    
    def learn(self,states,actions,reward):
        if states==[0,1,0] or states==[0,0,1]:
            self.Q[1,0] = self.Q[1,0] + self.lr * (reward - 1 - self.Q[1,0])
        else :
            self.Q[0,0] = self.Q[0,0] + self.lr * (reward - self.Q[0,0])
        if actions==1:
            if states==[0,1,0]:
                self.Q[1,1] = self.Q[1,1] + self.lr*(reward-self.Q[1,1])      
            elif states==[0,0,1]:
                self.Q[1,2] = self.Q[1,2] + self.lr*(reward-self.Q[1,2])    
            else:
                self.Q[1,1] = self.Q[1,1] + 0.1*self.lr*(reward-self.Q[1,1])      
                self.Q[1,2] = self.Q[1,2] + 0.1*self.lr*(reward-self.Q[1,2])     

def log_P_rl(q0,lr,Stay_cue,Safe_risk,Safe_risk_0,No_ask,Can_ask):
    rl = RL_agent(q0,lr)
    log_p = 0
    for i in range(120):
        if Can_ask == 1:
            log_p += rl.P_action_softmax([1,0,0],No_ask[i])
        if Stay_cue[i]==0:
            state = [0,0.5,0.5]
            log_p += np.log(rl.P_action_con0()[Safe_risk_0[i]])
        elif Stay_cue[i]==1:
            state = [0,1,0]
            log_p += np.log(rl.P_action_con1()[Safe_risk_0[i]])
        elif Stay_cue[i]==2:
            state=[0,0,1]
            log_p += np.log(rl.P_action_con2()[Safe_risk_0[i]])
        else:
            print('ERROR')

        rl.learn(state,Safe_risk_0[i],Safe_risk[i])
    return log_p

def Parameter_update_rl(Stay_cue,Safe_risk,Safe_risk_0,No_ask,Can_ask):
    GAMMA = np.linspace(0.01,1.5,100)
    LR = np.linspace(0.01,1.00,99)
    log_p = log_P_rl(0,0.1,Stay_cue,Safe_risk,Safe_risk_0,No_ask,Can_ask)
    log_p0 = -500
    for i in GAMMA:
        for j in LR:
            log_p = log_P_rl(i,j,Stay_cue,Safe_risk,Safe_risk_0,No_ask,Can_ask)
            if log_p>log_p0:
                log_p0 = log_p
    return log_p            
```
